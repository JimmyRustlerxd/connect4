```python
import numpy as np
from IPython.display import clear_output
import random
```

```python
def playmove(b,i,j,a):
    b[j[i],i] = a
    j[i] += 1
    return b,j,a*-1
```

```python
def validmove(i,j):
    if (i < wid):
        if (j[i] < hei) :
            return True

    return False
```

```python
def check4(b,i,j):
    if j < wid-3:
        if abs(sum(b[i,j:j+4])) == 4:
            return True
    if i < hei-3:
        if abs(sum(b[i:i+4,j])) == 4:
            return True       
    if i < hei-3 and j < wid-3:
        if abs(np.sum(b[i:i+4, j:j+4].diagonal())) == 4:
            return True
    if i >= 3 and j < wid-3:
        if abs(np.sum(np.fliplr(b[i-3:i+1, j:j+4]).diagonal())) == 4:
            return True
    return False
```

```python
def checkboard(b):
    for j in range(wid):
        for i in range(hei):
            if board[i,j] == 0:
                break
            if check4(b,i,j):
                return True
    return False
```

```python
def getmove(p):
    if p == 1:
        return int(input("Enter move"))
    if p == 2:
        return rand()
    if p == 3:
        return rec1(board,top,active,6)[1]
```

```python
def rand():
    return random.randint(0,wid)
```

```python
def rec1(b, t, player, depth):
    if depth == 0 or (sum(t) >= hei*wid):
        return 0, None

    best = -float('inf')
    move = None

    for i in range(wid):
        if t[i] < hei:
            j = t[i]

            b[j,i] = player
            t[i] += 1

            if check4(b,j,i):
                score = 100+depth
            else:
                score = -rec1(b,t,-player,depth-1)[0]

            b[j,i] = 0
            t[i] -= 1

            if score > best:
                best = score
                move = i

    return best, move
```

```python
#Settings
wid = 9
hei = 7
#1 = Human
#2 = Random
p1 = 1
p2 = 3
```

```python
board = np.zeros((hei,wid),int)
top=np.zeros(wid,int)
active = 1
print(board[::-1])
while (sum(top) < hei*wid) and checkboard(board) == False:
    if active == 1:
        move = getmove(p1)
    else:
        move = getmove(p2)
    print(move)
    if validmove(move,top):
        board, top, active = playmove(board, move, top, active)
    else:
        print("Invalid Move")
    clear_output()
    print(board[::-1])
```
